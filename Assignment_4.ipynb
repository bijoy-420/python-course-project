# 1. Create a NumPy array with 20 random integers between 1 and 100 and its basic operations

```python
random_array=np.random.randint(1,100,20)
print(f'\n Random Array')
print(random_array)
print(f'\n Mean of Random Array:',np.mean(random_array))
print(f'\n The sum of Random Array:',np.sum(random_array))
print(f'\n The standard deviation of Random Array:',np.std(random_array))
print(f'\n Max Number of random Array:',np.max(random_array))
print(f'\n Min Number of random Array:',np.min(random_array))
```
