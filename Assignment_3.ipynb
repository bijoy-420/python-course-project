**4.Write a Python program that takes the user’s age and prints which age group they belong to:**
- 'Child' for age 0–12
- 'Teenager' for age 13–19
- 'Adult' for age 20–64
- 'Senior' for age 65 and above

```python
age =int(input("Enter your age:"))
if age>=0 and age<=12:
  print("Child")
elif age >=13 and age<=19:
  print("Teenager")
elif age>=20 and age<=64:
  print('Adult')
else:
  print("Senior")
```

**5.Write a Python function `sum_of_natural_numbers(n)` that returns the sum of all natural numbers up to `n`. Use a loop inside the function.**

```python
def sum_of_natural_numbers(n):
  sum=0
  for i in range (1,n+1):
    sum+=i
  return sum
```

```python
sum_of_natural_numbers(10)
```

**6.Write a Python function `factorial(n)` that calculates and returns the factorial of `n`. Use a loop to calculate the factorial.**

```python
def factorial(n):
  fact=1
  for i in range(1,n+1):
    fact=fact*i

  return fact
```

```python
factorial(5)
```

**7.Write a Python function `multiplication_table(n)` that prints the multiplication table of `n` up to 10**

```python
def multiplication_table(n):
  for i in range(1,11):
    print(f"{i}*{n}={n*i}")
```

```python
multiplication_table(3)
```
