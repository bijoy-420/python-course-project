# **#Calculate the area and perimeter of a rectangle with length 7 and width 5.**

# **#Create two sets of colors and find their union and intersection**

```python
color_1={'red','green','blue','black'}
color_2={'brown','white','pink','gray'}
union=color_1.union(color_2)
intersection=color_1.intersection(color_2)
print(f"union is ",union)
print(f'intersection is ',intersection)
```
